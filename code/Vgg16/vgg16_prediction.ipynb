```python
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
import numpy as np
import matplotlib.pyplot as plt
import os

# 1️⃣ Load the model (use your saved VGG16 model name)
model_path = "vgg16_full_model.h5"   # change if your name is different
model = load_model(model_path)
print("Model loaded successfully!")

# 2️⃣ Labels (must match training order)
labels = {'A+': 0, 'A-': 1, 'AB+': 2, 'AB-': 3,
          'B+': 4, 'B-': 5, 'O+': 6, 'O-': 7}

# Reverse dictionary
labels = {v: k for k, v in labels.items()}

# 3️⃣ Test Image Path
img_path = r"D:\fingerprint-based-blood-group-detection-main\dataset\dataset_blood_group\A+\cluster_0_1001.BMP"

if not os.path.exists(img_path):
    raise FileNotFoundError("Image not found: " + img_path)

# 4️⃣ Load & Preprocess Image (VGG16 uses 256x256)
img = image.load_img(img_path, target_size=(256, 256))
x = image.img_to_array(img)
x = np.expand_dims(x, axis=0)
x = x / 255.0

# 5️⃣ Predict
result = model.predict(x)
predicted_class = np.argmax(result)
predicted_label = labels[predicted_class]
confidence = result[0][predicted_class] * 100

# 6️⃣ Show Output
plt.imshow(img)
plt.axis("off")
plt.title(f"Prediction: {predicted_label} ({confidence:.2f}%)")
plt.show()
```
